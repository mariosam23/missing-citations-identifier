# Missing-citation detection — evaluation figures

Plot-only notebook. Figures are generated by `python -m scripts.build_eval_figures` and embedded from `../data/eval/figures/`. Re-run that script to refresh them.

```
from pathlib import Path
from IPython.display import Image, display

FIG_DIR = Path('../data/eval/figures')
```

## Rule detector confusion matrix

The rule detector almost never flags MISSING_CITATION (recall ~0.05) but never re-flags an already-cited sentence.

```
display(Image(filename=str(FIG_DIR / '01_rule_confusion_matrix.png')))
```

## LLM identifier confusion matrix

The LLM recovers most should-cite sentences; remaining false positives are largely weak-label noise.

```
display(Image(filename=str(FIG_DIR / '02_llm_confusion_matrix.png')))
```

## Rule vs LLM identification

Switching the decision to the LLM lifts should-cite recall from 0.05 to 0.84 (~17x) at comparable precision.

```
display(Image(filename=str(FIG_DIR / '03_rule_vs_llm.png')))
```

## Confidence threshold sweep

Self-reported confidence is a weak knob: metrics are flat until ~0.9, where a small precision/recall trade appears.

```
display(Image(filename=str(FIG_DIR / '04_threshold_sweep.png')))
```

## Input-sanitization filter — runs comparison

Two filter-OFF runs differ by ~0.16 recall, so free-tier model-rotation non-determinism dominates: the filter's headline-metric effect on this clean eval set is within that noise. Its deterministic value is residue removal, label-leak removal, and skipping non-prose targets.

```
display(Image(filename=str(FIG_DIR / '05_filter_ablation.png')))
```

## End-to-end staged recall

End-to-end recall = detector recall x recommender-on-detected. The LLM bar is projected from the measured LLM recall and the recommender's recall on detected sentences.

```
display(Image(filename=str(FIG_DIR / '06_end_to_end.png')))
```

## Recall by citation density

The recommender's recall climbs monotonically with how many corpus papers cite the gold paper — corpus density is the ceiling.

```
display(Image(filename=str(FIG_DIR / '07_density_stratified.png')))
```
